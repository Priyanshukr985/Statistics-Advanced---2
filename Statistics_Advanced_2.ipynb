**Question 1: What is hypothesis testing in statistics?**

Ans1 Hypothesis testing is a statistical method used to make decisions or inferences about a population based on sample data.
It involves:

Formulating two statements:

Null hypothesis (H₀): Assumes no effect or no difference.

Alternative hypothesis (H₁): Assumes there is an effect or difference.

Using sample data and a test statistic to determine whether there is enough evidence to reject the null hypothesis at a chosen significance level (α).

**Question 2: What is the null hypothesis, and how does it differ from the alternative hypothesis?**

Ans2 **Null Hypothesis (H₀):**

The null hypothesis represents the default assumption or no effect/no difference in the population.

It is what we try to test against using sample data.

Example: “The average daily sales of the store is 250 units.”

**Alternative Hypothesis (H₁ or Ha):**

The alternative hypothesis represents the claim we want to test for — there is an effect or difference.

It is considered if the null hypothesis is rejected.

Example: “The average daily sales of the store is not 250 units.”

**Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.**

Ans3 **Significance Level (α):**

The significance level, denoted by α, is the probability of rejecting the null hypothesis (H₀) when it is actually true.

In other words, it represents the maximum risk of making a Type I error.

**Common Values:**

0.05 (5%) → 95% confidence

0.01 (1%) → 99% confidence

**Role in Hypothesis Testing:**

Set α before performing the test.

Compare the p-value (probability of observing data as extreme as the sample) with α:

If p-value ≤ α → Reject H₀ (evidence supports H₁)

If p-value > α → Fail to reject H₀ (not enough evidence)

**Intuition:**

Smaller α → stricter test → less chance of false positive

Larger α → more lenient test → higher chance of detecting an effect but more risk of Type I error

**Question 4: What are Type I and Type II errors? Give examples of each.**

Ans4 **Type I Error (False Positive)**

Occurs when we reject the null hypothesis (H₀) even though it is true.

Probability: α (significance level)

Example: A retail company concludes that a new sales strategy increases sales, but in reality, it doesn’t.

**Type II Error (False Negative)**

Occurs when we fail to reject the null hypothesis (H₀) even though the alternative hypothesis (H₁) is true.

Probability: β

Example: A retail company concludes that the new sales strategy does not increase sales, but in reality, it does.


**Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.**

Ans5 Z-Test vs T-Test

**Z-Test**


Population Standard Deviation (σ) is known

Sample Size (n):  Large (n ≥ 30)

Test Statistic: Follows Z (standard normal) distribution

Use Case:
Compare sample mean with population mean (σ known)

**T-Test**

Population Standard Deviation (σ) is unknown

Sample Size (n):  Small (n < 30)

Test Statistic: Follows t-distribution (Student’s t)

Use Case:
Compare sample mean with population mean or two sample means (σ unknown)

**When to Use Each:**

Z-Test:

You know the population standard deviation.

Sample size is large (central limit theorem applies).

Example: Testing if the average height of students differs from a known population mean, and σ is known.

T-Test:

Population standard deviation is unknown.

Sample size is small.

Example: Testing if the average daily sales of a small store differs from a target value when σ is unknown.

```python
'''Question 8: Write a Python script to simulate data from a normal distribution and
calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.
(Include your Python code and output in the code box below.)'''

import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Parameters
mean = 50       # population mean
std_dev = 5     # population standard deviation
n = 1000        # sample size

# Generate random data from normal distribution
data = np.random.normal(loc=mean, scale=std_dev, size=n)

# Sample statistics
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)

# 95% Confidence Interval for the mean
confidence = 0.95
z_score = stats.norm.ppf(0.975)  # two-tailed 95% CI
margin_of_error = z_score * (sample_std / np.sqrt(n))
ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print("Sample Mean:", sample_mean)
print("95% Confidence Interval: ({:.2f}, {:.2f})".format(ci_lower, ci_upper))

# Plot histogram
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title("Histogram of Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
''' Question 9: Write a Python function to calculate the Z-scores from a dataset and
visualize the standardized data using a histogram. Explain what the Z-scores represent
in terms of standard deviations from the mean.
(Include your Python code and output in the code box below.)'''

import numpy as np
import matplotlib.pyplot as plt

# Sample dataset
data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
        50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5]

# Function to calculate Z-scores
def calculate_z_scores(dataset):
    mean = np.mean(dataset)
    std = np.std(dataset, ddof=1)  # sample standard deviation
    z_scores = [(x - mean)/std for x in dataset]
    return z_scores

# Calculate Z-scores
z_scores = calculate_z_scores(data)

# Print first 10 Z-scores
print("First 10 Z-scores:", z_scores[:10])

# Plot histogram of Z-scores
plt.hist(z_scores, bins=10, color='lightcoral', edgecolor='black')
plt.title("Histogram of Z-scores")
plt.xlabel("Z-score")
plt.ylabel("Frequency")
plt.show()
```
